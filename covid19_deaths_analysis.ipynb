```python
import pandas as pd
url = 'https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/owid-covid-data.csv'
df3 = pd.read_csv(url)
df3.head()
```

```python
total_deaths = df3.groupby('continent')['total_deaths'].sum()
print(total_deaths)
```

```python
sorted_deaths = total_deaths.sort_values(ascending=False)
print(sorted_deaths)
```

```python
import matplotlib.pyplot as plt
plt.bar(sorted_deaths.index, sorted_deaths.values)
plt.title('Total COVID-19 Deaths by Continent')
plt.xlabel('Continent')
plt.ylabel('Total Deaths')
plt.show()
```

The findings demonstrate that COVID-19 was a global crisis, but its effects were unevenly distributed across continents. The large differences in recorded deaths highlight the importance of public health preparedness, healthcare capacity, demographic characteristics, and government response strategies in shaping pandemic outcomes. These results underscore the need for continued investment in disease surveillance, healthcare systems, and international cooperation to better manage future global health emergencies.
